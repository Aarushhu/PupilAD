```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load all four tables
upenn = pd.read_csv('../data/adni/UPENNBIOMK_MASTER.csv')
dxsum = pd.read_csv('../data/adni/DXSUM.csv')
ptdemog = pd.read_csv('../data/adni/PTDEMOG.csv')
modhach = pd.read_csv('../data/adni/MODHACH.csv')

# Quick summary of each
for name, df in [('UPENNBIOMK', upenn), ('DXSUM', dxsum), 
                  ('PTDEMOG', ptdemog), ('MODHACH', modhach)]:
    print(f"\n{'='*40}")
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
    print(f"Columns: {list(df.columns)}")
```

```python
# Merge all tables on RID
merged = upenn[['RID', 'VISCODE', 'TAU', 'PTAU', 'ABETA']].merge(
    dxsum[['RID', 'VISCODE', 'DIAGNOSIS']], on=['RID', 'VISCODE'], how='left'
).merge(
    ptdemog[['RID', 'PTGENDER', 'PTDOB', 'PTEDUCAT']], on='RID', how='left'
).merge(
    modhach[['RID', 'VISCODE', 'HMSCORE']], on=['RID', 'VISCODE'], how='left'
)

print(f"Merged dataset: {merged.shape[0]} rows, {merged.shape[1]} columns")
print(f"\nMissing values:\n{merged.isnull().sum()}")
print(f"\nDiagnosis breakdown:\n{merged['DIAGNOSIS'].value_counts()}")
print(f"\nTAU stats:\n{merged['TAU'].describe()}")
```
